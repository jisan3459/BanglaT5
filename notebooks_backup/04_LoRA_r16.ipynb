```python
from google.colab import drive
drive.mount("/content/drive")

from pathlib import Path
PROJECT_ROOT = Path("/content/drive/MyDrive/BanglaT5_Project")

if not PROJECT_ROOT.exists():
    raise FileNotFoundError(
        f"Project folder not found: {PROJECT_ROOT}. "
        "Make sure the extracted folder is directly inside MyDrive."
    )

print("Project root:", PROJECT_ROOT)
```

```python
import subprocess
import sys

# torchao is not used in this study. Removing an old preinstalled build avoids
# the PEFT compatibility error seen in some Colab runtimes.
subprocess.run(
    [sys.executable, "-m", "pip", "uninstall", "-y", "torchao"],
    check=False,
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL,
)

requirements_file = PROJECT_ROOT / "requirements" / "lora.txt"
subprocess.check_call([
    sys.executable, "-m", "pip", "install", "-q", "-r", str(requirements_file)
])
print("LoRA training dependencies installed.")
```

```python
import torch
import codecarbon
import peft

print("CUDA available:", torch.cuda.is_available())
if not torch.cuda.is_available():
    raise RuntimeError("Enable a T4 GPU before training.")
print("GPU:", torch.cuda.get_device_name(0))
print("CodeCarbon package:", codecarbon.__file__)
print("PEFT version:", peft.__version__)

if str(PROJECT_ROOT) in str(codecarbon.__file__):
    raise RuntimeError(
        "A local folder is shadowing the installed CodeCarbon package. "
        "Use this fixed project package, which stores logs in carbon_logs/."
    )
```
